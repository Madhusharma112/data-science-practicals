```python
# STUDENT RATING DATASET

import pandas as pd
import matplotlib.pyplot as pyplot

# Load dataset
ratings_df = pd.read_csv("C:/Users/Ajsha/Downloads/teachingratings.csv")

print("Dataset loaded successfully!")
print("Number of observations:", len(ratings_df))

print("\nFirst 5 rows:")
print(ratings_df.head())
```

```python
# Q1. VISIBLE MINORITY AND TENURE

minority_professors = ratings_df[ratings_df["minority"] == "yes"]

minority_tenured = (
    minority_professors["tenure"] == "yes"
).mean() * 100

nonminority_tenured = (
    ratings_df[ratings_df["minority"] == "no"]["tenure"] == "yes"
).mean() * 100

print("\nQ1. TENURE STATUS BY MINORITY")

print("Percentage of visible-minority professors who are tenured:",
      round(minority_tenured, 2), "%")

print("Percentage of non-visible-minority professors who are tenured:",
      round(nonminority_tenured, 2), "%")
```

```python
# Q2. AGE BY TENURE

age_tenure = ratings_df.groupby("tenure")["age"].agg(
    ["mean", "std"]
).reset_index()

print("\nQ2. AGE BY TENURE")

print(age_tenure.round(2))
```

```python
# Q3. HISTOGRAM OF AGE

print("\nQ3. HISTOGRAM OF AGE")

pyplot.hist(
    ratings_df["age"],
    bins=10,
    color="purple",
    edgecolor="black"
)

pyplot.xlabel("Age")
pyplot.ylabel("Frequency")
pyplot.title("Distribution of Professor Age")

pyplot.show()
```

```python
# Q4. GENDER BAR GRAPH

print("\nQ4. GENDER BAR GRAPH")

gender_count = ratings_df["gender"].value_counts()

pyplot.bar(
    gender_count.index,
    gender_count.values,
    color="teal",
    edgecolor="black"
)

pyplot.xlabel("Gender")
pyplot.ylabel("Count")
pyplot.title("Gender Distribution")

pyplot.show()
```

```python
# Q5. MEDIAN EVALUATION SCORE FOR TENURED PROFESSORS

tenured_professors = ratings_df[
    ratings_df["tenure"] == "yes"
]

median_eval = tenured_professors["eval"].median()

print("\nQ5. MEDIAN EVALUATION SCORE")

print("Median evaluation score for tenured professors:",
      median_eval)
```
