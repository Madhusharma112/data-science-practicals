```python
import pandas as pd
import numpy as np
# Step 1:RAW DATA 

data = {
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Midhat', 'Mahender', 'Kritika', 'Ajay', 'Shruti', 'Ritik'],
    'Age': [21, 22, 20, 23, 21, 22],
    'Study_Hours': [5, 7, 4, 8, 6, 3],
    'Marks': [72, 85, 65, 91, 78, 59],
    'Department': ['IT', 'CS', 'IT', 'DS', 'CS', 'IT']
}
df = pd.DataFrame(data)
print("raw data")
print(df)
print("\nDATAFRAME INFORMATION")
df.info()
# Step 2: NUMPY - SHAPE, SIZE, DIMENSION
numeric_df = df[
    ['Age', 'Study_Hours', 'Marks']
]

arr = numeric_df.to_numpy()

print("\nNUMPY ARRAY")
print(arr)

print("\nShape:", arr.shape)
print("Size:", arr.size)
print("Dimension:", arr.ndim)
# Step 3: NUMPY OPERATIONS ON TWO ARRAYS

A = np.array([
    [2, 3],
    [4, 5]
])

B = np.array([
    [6, 7],
    [8, 9]
])

print("\nARRAY A:")
print(A)

print("\nARRAY B:")
print(B)

# Addition
print("\nADDITION")
print(A + B)

# Subtraction
print("\nSUBTRACTION")
print(A - B)

# Multiplication
print("\nMULTIPLICATION")
print(A * B)
# Matrix Multiplication
print("\nMATRIX PRODUCT")
print(A @ B)

# Transpose
print("\nTRANSPOSE OF A")
print(A.T)

print("\nTRANSPOSE OF B")
print(B.T)

# Inverse
print("\nINVERSE OF A")
print(np.linalg.inv(A))

print("\nINVERSE OF B")
print(np.linalg.inv(B))
# Step 4: ONE-HOT ENCODING
print("\nCATEGORICAL DATA")
print(df[['Name', 'Department']])
encoded_df = pd.get_dummies(
    df,
    columns=['Department'],
  dtype=int
)
print("\nOne-Hot Encoded Data:")
print(encoded_df)
```
